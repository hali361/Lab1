## Error test case

**input:** 190, 11/24/2002, Darin Vale,, 1.58

**expected behavior**: 'MissingValueException' should be raised as there is a missing value under the weight category.

```python
error_row = "190, 11/24/2002, Darin Vale,,1.58"

try:
    parse_row(error_row)

except MissingValueException:
    print("Input:", error_row)
    print("MissingValueException has been raised.")
```
